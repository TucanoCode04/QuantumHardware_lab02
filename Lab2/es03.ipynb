```python
from qiskit import QuantumCircuit, QuantumRegister
import matplotlib.pyplot as plt

names = ["a0", "a1", "b0", "b1", "s0", "s1", "s2"]

qc = QuantumCircuit(*[QuantumRegister(1, name) for name in names])

qc.x(0)
qc.mcx([0, 1, 2, 3], 4)
qc.x(2)
qc.mcx([0, 1, 2, 3], 5)

qc.x(4)
qc.x(5)
qc.ccx(4, 5, 6)
qc.x(6)

fig = qc.draw(output="mpl", plot_barriers=False)
display(fig)
plt.close(fig)
```
